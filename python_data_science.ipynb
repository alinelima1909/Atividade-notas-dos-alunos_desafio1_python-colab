```python
num1 = float(input('Digite o 1° número: '))
num2 = float(input('Digite o 2° número:  '))

if num1 > num2:
  print( f' O primeiro número é maior:{num1}')
elif num2 > num1:
  print(f'O segundo número é maior:{num2}')
else:
  print(f' Os dois  números são iguais.')
```

```python
# Coletar os números
num1 = float(input('Digite o 1° número: '))
num2 = float(input('Digite o 2° número: '))

# Comparamos ambos os números e descobrimos qual é o maior
if num1 > num2:
    print(f'O primeiro número é maior: {num1}')
elif num2 > num1:
    print(f'O segundo número é maior: {num2}')
else: # Caso os números sejam iguais
    print('Os dois números são iguais.')
```

```python
# Coletar do percentual
variacao = float (input('Digite o percentual de crescimento: '))

# Verificar se o valor é positivo ou negativo com uma verificação se o numero
# é maior ou menor que 0
if variacao > 0:
  print(f' Houve um crescimento de {variacao}%')
elif variacao < 0:
  print(f' Houve um decrescimento de {variacao}%')
else:
  print('Nao Houve crescimento ou decrescimento')
```

```python
# Coletamos a letra da pessoa usuária como minúscula
letra = input('Digite uma letra: ').lower()
vogais = 'aeiou' # string contendo todos os dados

# Verificamos se a letra está nas vogais com in
if letra in vogais:
    print('A letra é uma vogal.')
else:
    print('A letra é uma consoante.')
```

```python
# Coletmaos os preços dos 3 anos
preco_ano1 = float(input('Digite o preço do primeiro ano:'))
preco_ano2 = float(input('Digite o preço do segundo ano:'))
preco_ano3 = float(input('Digite o preço do terceiro ano:'))

# Determinamos o maior valor entre as comparações
maior = preco_ano1
if preco_ano2 > maior:
  maior = preco_ano2
if preco_ano3 > maior:
  maior = preco_ano3
# Determinamis i menor valor entre as comparações
menor = preco_ano1
if preco_ano2 < menor:
  menor = preco_ano2
if preco_ano3 < menor:
  menor = preco_ano3

# Mostramos o resultado
print(f'O preco mais alto foi de R$ {maior}.')
print(f'O preço mais baixo foi de R$ {menor}.')
```
