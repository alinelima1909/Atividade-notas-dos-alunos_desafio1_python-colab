# Notebook 01 – Preparação dos Dados

## Contexto
Este notebook tem como objetivo realizar a preparação dos dados que serão utilizados
nas análises de internações, reincidência, evasão de tratamento e perfil do paciente
em um projeto de Ciência de Dados aplicado à Saúde Mental.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import os
```

Padronização de nomes das colunas

```python
df.columns = (
    df.columns
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('-', '_')
)
```

```python
df.columns
```

Tipos unicos de colunas para inspecção

```python
for col in df.columns:
    print(f'\nColuna: {col}')
    print(df[col].unique()[:10])
```

Salvando a base de dados

```python
df.to_csv('/content/mental_health_dataset_preparado.csv', index=False)
```
